```python
import statistics
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

#Number of people in the set
num_people = len(ages)
print(f"Total people: {num_people}")

#Ages
total_age = sum(ages)
print(f"Total of all ages: {total_age}")

#Mean age to 1d.p
mean_age = total_age / num_people
print(f"Mean age: {mean_age:.1f}")

#highest and lowest age
youngest = min(ages)
oldest = max(ages)
print(f"Youngest age: {youngest}, Oldest age: {oldest}")

median_age = statistics.median(ages)
print(f"Median age: {median_age}")
```

1. Student ID Number / Sequential Registration Number
Why it doesn't help: This is an arbitrary, system-generated identifier used strictly for administrative tracking, security, and record-keeping. The digits assigned to a student carry no information about their prior academic performance, study habits, motivation, or cognitive ability.

2. Dorm Room / Housing Assignment
Why it doesn't help: Specific physical attributes like room number, building floor, or bed location are administrative outcomes—often assigned randomly or based on general availability—rather than indicators of academic capability. While overall living conditions (like stress or noise) can impact a student, the specific room assignment data point lacks any intrinsic correlation with how well a student grasps coursework or performs on assessments.

3. Demographic / Socioeconomic Class Status
Why it doesn't help: While broad demographic data is collected for equity and reporting purposes, meta-analyses on admitted college populations show that general demographic variables (like family income bracket or geographic origin) have little to no direct predictive power for an individual student's GPA. Because admitted students have already gone through a selective admissions process, personal academic behaviors—such as time management, course engagement, and self-efficacy—overwhelmingly drive academic outcomes rather than their demographic background.

```python
import pandas as pd
#load the file
df = pd.read_excel("50_startups_messy.xlsx")

#ampount and number of rows
print("Shape",df.shape)

#missing count per coloumn
print("Missing counts: /n", df.isnull().sum())

#to report duplicate count
print("Duplicate count: ", df.duplicated().sum())
```

```python
df_clean = df.drop_duplicates().copy()
#Standardise state column ('new york' -> 'New York')
df_clean["State"] = df_clean["State"].str.strip().str.title()

df_clean["Marketing Spend"] = pd.to_numeric(
    df_clean["Marketing Spend"],errors="coerce"
)
```

```python
srs_sample = df_clean.sample(n=20, random_state=7)

strat_sample = df_clean.groupby("State",group_keys=False).apply(
    lambda x: x.sample(frac=0.20, random_state=7)
)
```

The Simple Random Sample has a bigger size of sample and already equal groups which makes it closer to the true population average for R&D and Marketing spend
